```python
#리스트내포-리스트를 줄여쓰기
[결과로 담을 값 for 항목 in 원본리스트] :원본에서 하나씩 꺼내라
```

```python
#일반 리스트문
numbers = [1, 2, 3, 4, 5]
result = []

for x in numbers:
    result.append(x * 2)

print(result)
```

```python
#리스트내포문
numbers = [1, 2, 3, 4, 5]

result = [x * 2 for x in numbers]

print(result)
```

```python
#리스트내포문/if뒤
numbers = [1, 2, 3, 4, 5]

result = [x for x in numbers if x % 2 == 0]

print(result)
```

```python
#리스트내포문/if앞
numbers = [1, 2, 3, 4, 5]

result = ["짝수" if x % 2 == 0 else "홀수" for x in numbers]

print(result)
```
