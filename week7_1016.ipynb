```python
#1부터 100까지 합을 출력하시오
sum = 0
for i in range(1,101):
  sum = sum+ i
print(sum)
```

```python
#1부터 100까지의 짝수의 합을 range()의 증가값을 이용하여 구하세요
#if 없이
sum = 0
for i in range(0, 101, 2):
  sum = sum + i
print(sum)
```

```python
#if 있게
sum = 0
for i in range(101):
  if (i%2 == 0):
    sum += i
print(sum)
```

```python
#정수 하나를 입력받아 그 수의 구구단을 7X1=7 형식으로 출력하세요
num = int(input("정수를 입력하세요:"))

for i in range(1,10):
  total = num * i
  print(num,"X", i, "=",total)

```

```python
#2단부터 9단까지 구구단을 7X1=7 형식으로 출력하세요

for i in range(2, 10):
  for j in range(1,10):
    total=i *j
    print(i,"X",j,"=", total)
  print("-------------------------")
```

```python
#while문으로 1부터 100까지 합을 출력하시오
sum = 0;
i=1
while(i<=100):
  sum +=i
  i+= 1
print(sum)
```

```python
#while문으로 1부터 100까지 합을 출력하시오
sum = 0;
i=1
while True:
  sum +=i
  i+= 1

  if (i>100):
    break
print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요

sum = 0

while True:
  num =int(input("양의 정수를 입력하세요"))
  if num == 0:
    break
  sum += num

print(sum)
```

```python
#1~20까지의 수 중 짝수만 담은 리스트를 만들어 출력하세요

evens = []

for i in range(1,21):
  if i%2 == 0:
    evens.append(i)
print(evens)

evens.pop(7)
print(evens)

del evens[5]
print(evens)

print(evens[3])
print(evens[-1])  #리스트에서 마지막의 것
```

```python
total = 0
scores =[ 85, 95, 70, 75, 99, 93, 88]

#scores의 항목이 들어옴
for i in scores:
  total += i
print(total)
```

```python
max_score = 0
scores =[ 85, 95, 70, 75, 99, 93, 88]

for i in scores:
  if (i > max_score):
    max_score = i
print(max_score)
```
