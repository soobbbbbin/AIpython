```python
#두 자연수 10와 5가 주어진다.
#이때, A+B, A-B, A*B, A/B(몫),A%B(나머지)를 출력하는 프로그램을 작성하시오.
A = 10
B = 5
print(A+B)
print(A-B)
print(A*B)
print(A/B)
print(A%B)
```

```python
#가로 12, 세로 7인 직사각형의 넓이와 둘레를 구하시오
w= 12
h= 7
s = w * h
p = 2 * (w + h)
print(s)
print(p)
```
