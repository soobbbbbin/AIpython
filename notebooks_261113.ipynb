```python
import numpy as np
import matplotlib.pyplot as plt

math = np.array([75, 89, 75, 65, 79])
english = np.array([85, 80, 90, 75, 88])
korean = np.array([72, 90, 70, 88, 93])
x = ['math', 'english', 'korean']

plt.figure(figsize=(5, 3))
plt.bar(x, [np.mean(math), np.mean(english), np.mean(korean)])
plt.ylabel('score')
plt.show()
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 01 - 넘파이 배열 만들기와 요소 접근 (코드 9-1, 9-2)

array 함수로 리스트를 배열로 바꾸고, 리스트와 같은 방법으로 인덱싱·슬라이싱한다.
"""
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print('-- (1) 배열 출력 (코드 9-1) --')
print(arr)
print(type(arr))

print()
print('-- (2) 인덱싱과 슬라이싱 (코드 9-2) --')
print(arr[1])        # 인덱싱
print(arr[1:4])      # 슬라이싱
print(arr[-1])       # 음수 인덱스도 리스트와 같다
print(arr[::2])      # 증가값도 같다

print()
print('-- (3) 리스트와 배열의 출력 모양 비교 --')
lst = [1, 2, 3, 4, 5]
print(lst)           # 쉼표가 있다
print(arr)           # 쉼표가 없다
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 02 - 배열을 만드는 여러 가지 함수 (표 9-1)

arange, linspace, zeros, ones, full
"""
import numpy as np

print('-- (1) arange: 간격이 일정한 배열 (range와 같은 규칙) --')
arr1 = np.arange(1, 8, 1)
print(arr1)
print(np.arange(5))            # 0 ~ 4
print(np.arange(10, 0, -3))    # 증가값이 음수

print()
print('-- (2) linspace: 요소 개수를 지정 --')
arr2 = np.linspace(1, 9, 5)
print(arr2)                    # 끝 값(9)을 포함한다

print()
print('-- (3) zeros, ones, full --')
print(np.zeros(10))
print(np.ones(10))
print(np.full(10, 5))

print()
print('-- (4) arange 와 linspace 의 차이 --')
print(np.arange(0, 1, 0.25))       # 끝 값 1 은 포함되지 않는다
print(np.linspace(0, 1, 5))        # 끝 값 1 이 포함된다
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 03 - 2차원 배열과 shape (코드 9-3, 9-4, 9-5)"""
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])

print('-- (1) 2행 3열 배열 (코드 9-3) --')
print(arr)

print()
print('-- (2) [행, 열] 로 요소 접근 (코드 9-4) --')
print(arr[0, 1], arr[1, 2])
print(arr[0][1], arr[1][2])    # arr[0][1] 로 써도 같다

print()
print('-- (3) shape: 각 차원의 크기 (코드 9-5) --')
print(arr.shape)
print('행의 수:', arr.shape[0])
print('열의 수:', arr.shape[1])

print()
print('-- (4) 행과 열 슬라이싱 --')
print(arr[0])          # 0행 전체
print(arr[:, 1])       # 1열 전체
print(arr[1, 1:])      # 1행의 1열부터
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 04 - reshape 로 배열의 모양 바꾸기 (코드 9-6, 표 9-2)"""
import numpy as np

arr1 = np.array([1, 2, 3, 4, 5, 6])
arr2 = arr1.reshape(3, 2)

print('-- (1) 1차원 -> 3행 2열 (코드 9-6) --')
print(arr1)
print(arr2)

print()
print('-- (2) 모양 확인 --')
print(arr1.shape)
print(arr2.shape)

print()
print('-- (3) 요소 접근 비교 (표 9-2) --')
print(arr1[3])         # 1차원은 인덱스 하나
print(arr2[1][1])      # 2차원은 [행][열] - 같은 값 4

print()
print('-- (4) 다른 모양으로 바꾸기 --')
print(arr1.reshape(2, 3))
print(arr1.reshape(2, 3).shape)
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 05 - 배열 연산 (표 9-3, 코드 9-7, 9-8)

리스트는 항목마다 일일이 계산해야 하지만 배열은 한 번에 계산된다.
"""
import numpy as np

print('-- (1) 리스트의 각 항목에 2 곱하기 (표 9-3) --')
lst = [2, 3, 4]
result = []
for i in range(len(lst)):
    result.append(lst[i] * 2)
print(result)
print([x * 2 for x in lst])      # 리스트 내포

print()
print('-- (2) 배열에 2 곱하기 (코드 9-7) --')
arr = np.array([2, 3, 4])
print(arr * 2)

print()
print('-- (3) 배열과 배열의 연산 (코드 9-8) --')
arr1 = np.array([2, 3, 4])
arr2 = np.array([5, 6, 7])
print(arr1 + arr2)
print(arr1 * arr2)

print()
print('-- (4) 주의: 리스트에 * 2 를 하면? --')
print(lst * 2)                   # 계산이 아니라 리스트가 반복된다
print(lst + [5, 6, 7])           # 더하기가 아니라 리스트가 이어 붙는다
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 06 - 이차함수 y = x^2 그래프 (코드 9-9)

linspace 로 x 배열을 만들고, x**2 로 y 배열을 한 번에 만든다.
"""
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 1, 100)
y = x**2

print('-- (1) 배열 확인 --')
print('x 의 요소 수:', len(x), ' y 의 요소 수:', len(y))
print('x 처음 3개:', x[:3])
print('y 처음 3개:', y[:3])
print('y 의 최솟값:', y.min())

plt.figure(figsize=(5, 3))
plt.plot(x, y, label='y=x^2')
plt.legend()
plt.show()
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 07 - xticks 로 x축 눈금 조절하기 (코드 9-10, 9-11)

우리나라 PISA 수학 성적 선 그래프.
year 를 그대로 x 로 쓰면 눈금이 2006, 2008, ... 로 나오므로,
arange 로 만든 ind 를 x 로 쓰고 xticks 로 눈금 이름을 year 로 바꾼다.
"""
import numpy as np
import matplotlib.pyplot as plt

year = [2006, 2009, 2012, 2015, 2018]
kor = [547, 546, 554, 524, 526]

print('-- (1) 눈금 위치로 쓸 배열 --')
ind = np.arange(len(year))
print(ind)

plt.figure(figsize=(10, 3))

plt.subplot(1, 2, 1)                 # 코드 9-10: 눈금이 어긋난다
plt.plot(year, kor)
plt.ylim(500, 580)
plt.title('(a) year as x')
plt.xlabel('year')
plt.ylabel('score')

plt.subplot(1, 2, 2)                 # 코드 9-11: 눈금을 맞춘다
plt.plot(ind, kor)
plt.xticks(ind, year)
plt.ylim(500, 580)
plt.title('(b) xticks')
plt.xlabel('year')
plt.ylabel('score')

plt.tight_layout()
plt.show()
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 08 - 막대 그래프가 겹치는 문제 (코드 9-12)

bar 를 두 번 호출하면 같은 위치에 그려져 남성 막대와 여성 막대가 겹친다.
"""
import numpy as np
import matplotlib.pyplot as plt

nation = ['Korea', 'USA', 'Japan', 'France']
men = [175.5, 176.9, 172.1, 178.6]
women = [163.2, 163.3, 158.5, 164.5]
ind = np.arange(len(nation))

print('-- (1) 막대의 위치 --')
print(ind)

plt.figure(figsize=(5, 3))
plt.bar(ind, men, color='b', label='men')
plt.bar(ind, women, color='r', label='women')
plt.ylim(130, 190)
plt.xticks(ind, nation)
plt.legend()
plt.show()
```

```python
# -*- coding: utf-8 -*-
"""CH09 LAB 10 - 누적 막대 그래프 (코드 9-14, 9-15)

bottom 매개변수로 막대의 시작 위치를 정한다.
bottom 에 spring+summer 처럼 덧셈을 쓰려면 spring, summer 가 배열이어야 한다.
"""
import numpy as np
import matplotlib.pyplot as plt

year = [2017, 2018, 2019, 2020, 2021]
spring = np.array([124.9, 383.5, 175.0, 173.7, 330.5])
summer = np.array([612.7, 620.6, 508.2, 1037.6, 612.8])
autumn = np.array([177.9, 351.3, 440.8, 270.4, 256.4])
winter = np.array([75.2, 68.7, 168.8, 47.8, 13.3])

print('-- (1) 연도별 연 강수량 --')
total = spring + summer + autumn + winter
for y, t in zip(year, total):
    print(f'{y}년: {t:.1f} mm')

print()
print('-- (2) 쌓이는 높이 (bottom 값) --')
print('summer 의 bottom :', spring)
print('autumn 의 bottom :', spring + summer)

plt.bar(year, spring, label='spring')
plt.bar(year, summer, bottom=spring, label='summer')
plt.bar(year, autumn, bottom=spring + summer, label='autumn')
plt.bar(year, winter, bottom=spring + summer + autumn, label='winter')
plt.title('precipitation')
plt.xlabel('year')
plt.ylabel('mm')
plt.legend()
plt.show()
```
