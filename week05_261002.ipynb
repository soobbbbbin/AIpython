```python
import random

com = random.randint(1, 100)
sum = 0

while True:
    user = int (input("수를 입력하세요"))
    sum = sum+1
    print(com)
    if user > com :
        print("그 수보다 작습니다")
    elif user < com :
        print("그 수보다 큽니다")
    else:
        print("정답입니다")
        print(f'{sum}번 시도했습니다')
        break
```

```python
# asmd 함수 정의

x = int (input ('첫 번째 수'))
y = int (input ('두 번째 수'))

op = input('연산자를 입력하시오')

def asmd(x, y ,op):
    if op == "+" :
        print(f'{x}+{y} == {x+y}')
    elif op == "-":
        print(f'{x}-{y} == {x-y}')
    elif op == "*":
        print(f'{x}*{y} == {x*y}')
    elif op == "/":
        print(f'{x}/{y} == {x/y}')
    else:
        print('잘못된 문자입니다')


        
        
        

    
```

```python
asmd(x, y ,op)
```
